```python
# Data Processing and Vectorization
import numpy as np
import pandas as pd
import pickle
from sentence_transformers import SentenceTransformer

# 1. This path is fixed in Kaggle's input directory, no need to change
input_path = '/kaggle/input/food-com-recipes-and-user-interactions/RAW_recipes.csv'

print("Reading data...")
# Read CSV
df = pd.read_csv(input_path)

# 2. Data cleaning and sampling (taking only 5000 records for lightweight processing)
# This step is very fast in the cloud environment
df = df.dropna(subset=['description', 'name', 'steps'])
df_sample = df.sample(n=5000, random_state=42).reset_index(drop=True)

# Clean tags format
df_sample['tags'] = df_sample['tags'].apply(lambda x: x.replace('[','').replace(']','').replace("'", ""))

# 3. Combine text
df_sample['combined_text'] = df_sample.apply(
    lambda row: f"Name: {row['name']}. Description: {row['description']}. Tags: {row['tags']}", 
    axis=1
)

print(f"Sampling completed, dataset size: {len(df_sample)}")

# 4. Vectorization (this will utilize Kaggle's CPU/GPU resources)
print("Loading model and performing vectorization...")
model = SentenceTransformer('all-MiniLM-L6-v2')
embeddings = model.encode(df_sample['combined_text'].tolist(), show_progress_bar=True)

# 5. Save processed data to Kaggle's Output directory
output_data = {
    "df": df_sample, 
    "embeddings": embeddings
}

# Save as pickle file
with open('recipe_data_with_embeddings.pkl', 'wb') as f:
    pickle.dump(output_data, f)

print("All done! Please check the Output section to download the file.")
```
